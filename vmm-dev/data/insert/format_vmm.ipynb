```python
# Format VMM MJP CSV exports for insert_vmm.ipynb.
# Run the cells in order.
# The notebook reads vmm-data, preserves valid hierarchy roots, and writes the JSONL
# handoff consumed by insert_vmm.ipynb.
```

```python
from __future__ import annotations

import json
from pathlib import Path
from typing import Any

import pandas as pd


def find_project_root() -> Path:
    candidates = [Path.cwd(), Path.cwd() / "vmm-dev"]
    candidates.extend(Path.cwd().parents)
    for candidate in candidates:
        candidate = candidate.resolve()
        if (candidate / "vmm-data").is_dir() and (candidate / "data" / "insert").is_dir():
            return candidate
    raise FileNotFoundError(
        "Could not find the vmm-dev project root. Run the notebook from vmm-dev "
        "or one of its child directories."
    )


PROJECT_ROOT = find_project_root()
INPUT_DIR = PROJECT_ROOT / "vmm-data"
OUTPUT_FILE = PROJECT_ROOT / "data" / "insert" / "data" / "vap_mjp_structured_2.jsonl"

CSV_FILES = {
    "aanleveringen": "aanlevering_MJP_2026.csv",
    "strategische_doelstellingen": "strategische_doesltelling_MJP_2026.csv",
    "beleidsdoelstellingen": "beleidsdoelstelling_MJP_2026.csv",
    "actieplannen": "actieplan_MJP_2026.csv",
    "acties": "actie_MJP_2026.csv",
}

print(f"Project root: {PROJECT_ROOT}")
print(f"CSV input:    {INPUT_DIR}")
print(f"JSONL output: {OUTPUT_FILE}")
```

```python
def clean(value: Any) -> Any:
    if pd.isna(value):
        return None
    if isinstance(value, pd.Timestamp):
        return value.isoformat()
    if hasattr(value, "item"):
        return value.item()
    return value


def identifier(value: Any) -> int | str | None:
    value = clean(value)
    if value is None or str(value).strip() == "":
        return None
    try:
        number = float(value)
        if number.is_integer():
            return int(number)
    except (TypeError, ValueError):
        pass
    return str(value).strip()


def boolean(value: Any) -> bool:
    value = clean(value)
    if value is None:
        return False
    if isinstance(value, bool):
        return value
    if isinstance(value, (int, float)):
        return bool(value)
    return str(value).strip().lower() in {"1", "true", "yes", "ja", "y", "t"}


def read_tables() -> dict[str, pd.DataFrame]:
    tables: dict[str, pd.DataFrame] = {}
    for table_name, filename in CSV_FILES.items():
        path = INPUT_DIR / filename
        if not path.is_file():
            raise FileNotFoundError(f"Required CSV file not found: {path}")
        tables[table_name] = pd.read_csv(path, dtype=object)
    return tables


def entity(row: pd.Series, fields: dict[str, str]) -> dict[str, Any]:
    return {name: clean(row[column]) for name, column in fields.items()}


def format_deliveries(tables: dict[str, pd.DataFrame]) -> tuple[list[dict[str, Any]], dict[str, int]]:
    deliveries: dict[int | str, dict[str, Any]] = {}
    skipped = {"beleidsdoelstellingen": 0, "actieplannen": 0, "acties": 0}

    for _, row in tables["aanleveringen"].iterrows():
        delivery_id = identifier(row["Aanlevering ID"])
        if delivery_id is None:
            raise ValueError("An aanlevering row is missing its ID.")
        if delivery_id in deliveries:
            raise ValueError(f"Duplicate aanlevering ID: {delivery_id}")
        deliveries[delivery_id] = {
            "id": delivery_id,
            "bestuur_id": identifier(row["Bestuur ID"]),
            "type_rapport": clean(row["Type rapport"]),
            "rapportjaar": identifier(row["Rapportjaar"]),
            "kwartaal": identifier(row["Kwartaal"]),
            "kredietjaar": identifier(row["Kredietjaar"]),
            "eindjaar": identifier(row["eindjaar"]),
            "meest_recente": clean(row["Meest recente"]),
            "status_rapport": clean(row["Status rapport"]),
            "toezicht_status": clean(row["Toezicht status"]),
            "presentatie_niveau": clean(row["Presentatie niveau"]),
            "oormerk": clean(row["Oormerk"]),
            "volgnummer": identifier(row["Volgnummer"]),
            "afgevlagd": clean(row["Afgevlagd"]),
            "datum_verwerking": clean(row["Datum verwerking"]),
            "datum_vaststelling": clean(row["Datum vaststelling"]),
            "datum_goedkeuring": clean(row["Datum goedkeuring"]),
            "strategische_doelstellingen": [],
            "actieplannen": [],
            "acties": [],
        }

    strategic_by_delivery: dict[tuple[int | str, int | str], dict[str, Any]] = {}
    policy_by_delivery: dict[tuple[int | str, int | str], dict[str, Any]] = {}
    actionplan_by_id: dict[int | str, tuple[int | str, dict[str, Any]]] = {}

    for _, row in tables["beleidsdoelstellingen"].iterrows():
        delivery_id = identifier(row["AANLEVERINGID"])
        policy_id = identifier(row["Beleidsdoelstelling ID"])
        if delivery_id not in deliveries or policy_id is None:
            skipped["beleidsdoelstellingen"] += 1
            continue
        sd_id = identifier(row["STRATEGISCHEDOELSTELLINGID"]) or 0
        strategic_key = (delivery_id, sd_id)
        strategic = strategic_by_delivery.get(strategic_key)
        if strategic is None:
            if sd_id == 0:
                strategic = {"id": 0, "code": "", "korte_omschrijving": "(geen strategische doelstelling)", "prioritair": False, "lange_omschrijving": None, "commentaar": None, "evaluatie": None, "dummy": None, "beleidsdoelstellingen": []}
            else:
                source = tables["strategische_doelstellingen"]
                matches = source[pd.to_numeric(source["Strategische doelstelling ID"], errors="coerce") == int(sd_id)]
                if matches.empty:
                    raise ValueError(f"Strategische doelstelling {sd_id} is not present in the source table.")
                source_row = matches.iloc[0]
                strategic = {
                    "id": sd_id,
                    "code": clean(source_row["Strategische doelstelling code"]),
                    "korte_omschrijving": clean(source_row["Strategische doelstelling korte omschrijving"]),
                    "prioritair": boolean(source_row["Strategische doelstelling prioritair"]),
                    "lange_omschrijving": clean(source_row["Strategische doelstelling lange omschrijving"]),
                    "commentaar": clean(source_row["Strategische doelstelling commentaar"]),
                    "evaluatie": clean(source_row["Strategische doelstelling evaluatie"]),
                    "dummy": clean(source_row["Strategische doelstelling dummy"]),
                    "beleidsdoelstellingen": [],
                }
            strategic_by_delivery[strategic_key] = strategic
            deliveries[delivery_id]["strategische_doelstellingen"].append(strategic)
        policy = entity(row, {
            "id": "Beleidsdoelstelling ID",
            "code": "Beleidsdoelstelling code",
            "korte_omschrijving": "Beleidsdoelstelling korte omschrijving",
            "prioritair": "Beleidsdoelstelling prioritair",
            "lange_omschrijving": "Beleidsdoelstelling lange omschrijving",
            "commentaar": "Beleidsdoelstelling commentaar",
            "evaluatie": "Beleidsdoelstelling evaluatie",
        })
        policy["id"] = policy_id
        policy["prioritair"] = boolean(row["Beleidsdoelstelling prioritair"])
        policy["actieplannen"] = []
        policy_by_delivery[(delivery_id, policy_id)] = policy
        strategic["beleidsdoelstellingen"].append(policy)

    for _, row in tables["actieplannen"].iterrows():
        delivery_id = identifier(row["AANLEVERINGID"])
        actionplan_id = identifier(row["Actieplan ID"])
        if delivery_id not in deliveries or actionplan_id is None:
            skipped["actieplannen"] += 1
            continue
        policy_id = identifier(row["BELEIDSDOELSTELLINGID"])
        actionplan = entity(row, {
            "id": "Actieplan ID",
            "code": "Actieplan code",
            "korte_omschrijving": "Actieplan korte omschrijving",
            "prioritair": "Actieplan prioritair",
            "lange_omschrijving": "Actieplan lange omschrijving",
            "commentaar": "Actieplan commentaar",
            "evaluatie": "Actieplan evaluatie",
        })
        actionplan["id"] = actionplan_id
        actionplan["prioritair"] = boolean(row["Actieplan prioritair"])
        actionplan["acties"] = []
        actionplan_by_id[actionplan_id] = (delivery_id, actionplan)
        policy = policy_by_delivery.get((delivery_id, policy_id)) if policy_id is not None else None
        if policy is not None:
            policy["actieplannen"].append(actionplan)
        elif policy_id is None:
            deliveries[delivery_id]["actieplannen"].append(actionplan)
        else:
            skipped["actieplannen"] += 1
            actionplan_by_id.pop(actionplan_id)

    for _, row in tables["acties"].iterrows():
        delivery_id = identifier(row["Aanlevering ID"])
        action_id = identifier(row["Actie ID"])
        if delivery_id not in deliveries or action_id is None:
            skipped["acties"] += 1
            continue
        action = entity(row, {
            "id": "Actie ID",
            "code": "Actie code",
            "korte_omschrijving": "Actie korte omschrijving",
            "prioritair": "Actie prioritair",
            "lange_omschrijving": "Actie lange omschrijving",
            "commentaar": "Actie commentaar",
            "evaluatie": "Actie evaluatie",
        })
        action["id"] = action_id
        action["prioritair"] = boolean(row["Actie prioritair"])
        actionplan_id = identifier(row["ACTIEPLANID"])
        parent = actionplan_by_id.get(actionplan_id) if actionplan_id is not None else None
        if parent is not None and parent[0] == delivery_id:
            parent[1]["acties"].append(action)
        elif actionplan_id is None:
            deliveries[delivery_id]["acties"].append(action)
        else:
            skipped["acties"] += 1

    return list(deliveries.values()), skipped


tables = read_tables()
deliveries, skipped = format_deliveries(tables)
print(f"Prepared {len(deliveries)} deliveries.")
print(f"  Strategic goals: {sum(len(d['strategische_doelstellingen']) for d in deliveries)}")
print(f"  Policy goals:    {sum(len(sd['beleidsdoelstellingen']) for d in deliveries for sd in d['strategische_doelstellingen'])}")
print(f"  Action plans:    {sum(len(bd['actieplannen']) for d in deliveries for sd in d['strategische_doelstellingen'] for bd in sd['beleidsdoelstellingen']) + sum(len(d['actieplannen']) for d in deliveries)}")
print(f"  Actions:         {sum(len(ap['acties']) for d in deliveries for sd in d['strategische_doelstellingen'] for bd in sd['beleidsdoelstellingen'] for ap in bd['actieplannen']) + sum(len(ap['acties']) for d in deliveries for ap in d['actieplannen']) + sum(len(d['acties']) for d in deliveries)}")
print(f"Skipped rows with unusable parent references: {skipped}")
```

```python
OUTPUT_FILE.parent.mkdir(parents=True, exist_ok=True)
with OUTPUT_FILE.open("w", encoding="utf-8") as handle:
    for delivery in deliveries:
        handle.write(json.dumps(delivery, ensure_ascii=False) + "\n")

print(f"Wrote {len(deliveries)} deliveries to {OUTPUT_FILE}")
```
